```python
import geopandas as gpd
import pandas as pd
import folium
import tomllib

from folium.plugins import Timeline, TimelineSlider
from folium.utilities import JsCode
```

```python
GEOJSON_FILE = "" # <-- ENTER YOUR GEOJSON OUTPUT FILE PATH HERE
```

```python
polys = gpd.read_file(GEOJSON_FILE)
polys = polys.to_crs("EPSG:4326")
polys["time"] = pd.Series([day.timestamp() * 1000 for day in list(pd.to_datetime(polys["time"]).dt.to_pydatetime())]).astype(int)
polys
```

```python
cfg = tomllib.load(open("app.toml", "rb"))
tower_location = cfg["input"]["location"]

map = folium.Map(
    location=tower_location,
    tiles=folium.TileLayer(
        tiles="http://mt0.google.com/vt/lyrs=s&hl=en&x={x}&y={y}&z={z}",
        attr="Google Maps"
    ),
    zoom_start=12
)

folium.Marker(
    location=tower_location
).add_to(map)

timeline = Timeline(
    polys.to_geo_dict(), 
    JsCode(
    '''
    function (daily) {
        return {
            start: daily.properties.time,
            end: daily.properties.time + 86000000,
        }
    }
    ''')
).add_to(map)

slider = TimelineSlider(
    auto_play=False,
    show_ticks=True,
    enable_keyboard_controls=True,
).add_timelines(timeline).add_to(map)

map.save(f"{cfg["name"]}_timeline.html")
```
